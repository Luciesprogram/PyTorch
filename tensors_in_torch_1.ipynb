## **Code to check to compare the time difference on GPU and CPU for matrix multiplication**

```python
import time

size = 10000 # matrix size

# create a random matrix on cpu
mat_1 = torch.rand((size,size))
mat_2 = torch.rand((size,size))

# Measure time on CPU
start_time = time.time()
result = torch.matmul(mat_1, mat_2)
end_time = time.time()
cpu_time = end_time - start_time

print(f"Time on CPU : {cpu_time: 4f} seconds")


# Moving a random matrix on GPU
matgpu_1 = mat_1.to("cuda")
matgpu_2 = mat_2.to("cuda")

# Measure time on CPU
start_time = time.time()
result_gpu = torch.matmul(matgpu_1, matgpu_2)
torch.cuda.synchronize()
gpu_time = time.time() - start_time

print(f"Time on GPU : {gpu_time: 4f} seconds")

# Compare Results
print("Speedup (cpu time/gpu time):",cpu_time / gpu_time)
```
